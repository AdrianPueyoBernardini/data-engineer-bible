## **Tuplas**

>**Basics**

```python
#Las tuplas son similares a las listas peeeero son inmutables. Los elementos no pueden ser reasignados.
mi_tupla =  (1,2,3)
mi_lista = [1,2,3]
mi_diccionario = {"clave1": "valor1", "clave2": "valor2"}
```

>**Función: count()**

```python
#Count nos indica cuantas veces se repite un valor dentro de la tupla
mi_tupla_2 =  (1,2,2,2,3)
print(mi_tupla_2.count(2))
```

>**Función: index()**

```python
#Index nos da el índice de la tupla
mi_tupla_2.index(3)
```
