## **Diccionarios**

>**Basics**

```python
#Diccionarios simples
#A diferencia de las listas, los diccionarios tienen un par de clave-valor
mi_dic =  {'clave1':'valor1', 'clave2':'valor2'}
print(mi_dic)

#A diferencia de las listas (mi_lista[0]) en el que buscamos por su key
#con los mapas podemos buscar por clave o valor indistintamente
print(mi_dic['clave2'])
```

>**Basics 2**

```python
#Los diccionarios son muy útiles, por ejemplo para emparejar por producto/precio, jugador/dorsal, etc...
mi_carrito = {"manzana":0.99, "pera":0.79, "sandía":2.99}
print(mi_carrito["manzana"])
```

>**Basics 3**

```python
#Los diccionarios pueden almacenar incluso otras listas dentro.
mi_dic_2 = {"k1":[0,1,2,3], "k2":[7,8,9,0], "k3": "valor3"}
print(mi_dic_2["k1"])
```

>**Basics 4**

```python
#Y por supuesto podemos modificar según índice.
mi_dic_3 = {"k1":[0,1,2,3], "k2":[7,8,9,0], "k3": ["a","b"]}
mi_dic_3["k3"][0].upper();

```

>**Funcion: .keys()**

```python
#Podemos obtener todas las claves del diccionario que queramos
mi_dic_3.keys()
```

>**Funcion: .values()**

```python
#También podemos obtener todos los valores del diccionario que queramos
mi_dic_3.values()
```
