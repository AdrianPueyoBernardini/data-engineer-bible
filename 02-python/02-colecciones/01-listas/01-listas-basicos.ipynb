## **Listas**

>**Basics**

```python
#Listas simples
mi_lista_num = [1,2,3,4]
mi_lista_str = ['a', 'b','c']
print(mi_lista_num)
print(mi_lista_str)

#Acceder al índice de la lista
print(mi_lista_num[0])
print(mi_lista_str[0])

#Cambiar elementos en un índice concreto
mi_lista_num[0] = 0
mi_lista_num[1] = 1
mi_lista_num[2] = 2
mi_lista_num[3] = 3
print(mi_lista_num)
```

>**Concatenación de Listas**

```python
#Concatenado simple
mi_lista_concat = mi_lista_num + mi_lista_str
print(mi_lista_concat)
```

>**Función: len()**

```python
#Listas simples
print(len(mi_lista_num))
```

>**Función: append()**

```python
#La función append nos sirve par añadir elementos a una lista
print(mi_lista_num) #Nuestra lista original

mi_lista_num.append(4) #Nuestra lista a la que le estamos agregando el valor 4 al final
print(mi_lista_num)
```

>**Función: pop()**

```python
#La función pop nos sirve para eliminar un elemento concreto de la tabla
print(mi_lista_num) #Nuestra lista original

mi_lista_num.pop() #eliminamos el último elemento
print(mi_lista_num)

mi_lista_num.pop(0) #también podemos eliminar el valor que queramos o incluso una string
print(mi_lista_num)

```

>**Función: sort()**

```python
#La función sort nos permite ordenar listas
#De esta forma ordenamos según alfabeto
mi_lista_desordenada = ['a', 'c', 'z', 'e', 'x','b', 'd']
print(mi_lista_desordenada)

mi_lista_desordenada.sort()
print(mi_lista_desordenada)

#También lo podemos hacer con valores numéricos
mi_lista_desordenada_2 = [5,2,1,3,4]
print(mi_lista_desordenada_2)

mi_lista_desordenada_2.sort()
print(mi_lista_desordenada_2)

```

>**Función: reverse()**

```python
#La función reverse nos permite invertir listas
print(mi_lista_desordenada_2)

mi_lista_desordenada_2.reverse()
print(mi_lista_desordenada_2)
```
