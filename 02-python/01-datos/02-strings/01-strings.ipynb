## **Strings**

>**Basics**

```python
#Strings
mi_str = 'Soy una cadena de caracteres'
mi_str_text = 'Soy un texto completo, creado, entre otras cosas para practicar' 
print(mi_str)
print(mi_str_text)
```

>**Formateo de texto**

*Saltos de linea* con **\n**

```python
#Podemos hacer saltos de linea con \n y concatenar Strings entre sí
mi_str_2 = 'Soy otra cadena de caracteres'
print('mi_str \nmi_str_2');
```

*Tabulaciones de linea* con **\t**

```python
#Podemos hacer tabulaciones de linea con \t 
print('mi_str \tmi_str_2');
```

>**Función: len**

```python
#Con la función len podemos saber el tamaño de la cadena, incluídos espacios
```

```python
len(mi_str)
```

>**String Index: String[0]**

```python
#Podemos obtener el index de los caracteres de la string que queramos, como si fueran Arrays
```

```python
print(mi_str)

#Index básico
print(mi_str[0]) #Nos devolverá S, la primera posición
print(mi_str[-1])#Nos devolverá S, la última posición

#Index combinados
print(mi_str[3:])#Nos devolverá a partir del valor 3
print(mi_str[:3])#Nos devolverá los 3 primeros valores
print(mi_str[8:14])#Nos devolverá entre el valor 8 y el 14

#Indices saltados
print(mi_str[::])#Nos devolverá toda la String
print(mi_str[::2])#Nos dará todos los valores saltando de 2 en 2
print(mi_str[2::2])#Nos dará todos los valores saltando de 2 en 2, podemos combinar para que inicie despues del 2

#Invertir nuestra cadena
print(mi_str[::-1])
```

>**Función: Lower() Y Upper()**

```python
mi_str = mi_str.upper() #Podemos convertir nuestra cadena a mayúsculas o minúsculas
print(mi_str)
mi_str.lower() #También podemos realizar esto a la inversa
```

>**Función: Split()**

```python
 #Split nos permite separar y convertir en un Array en base a las palabras
mi_str.split() #Separa por espacios
print(mi_str.split())

mi_str_text.split(',') #Separa por comas o cualquier caracter que queramos
print(mi_str_text.split(','))
```

>**Función: Format()**

```python
#Con format podemos insertar y concatenar caracteres
print('Ejemplo {}'.format('INSERTED'))

#Podemos hacer esto modular y añadir tantos brackets como queramos
print('Ejemplo {} {} {}'.format('Añadido1', 'Añadido2', 'Añadido3'))

#También podemos reorganizarlo como queramos
#Podemos hacer esto modular y añadir tantos brackets como queramos, por ejemplo a la inversa
print('Ejemplo {2} {1} {0}'.format('Añadido1', 'Añadido2', 'Añadido3'))

#Formateo de flotantes, con .2 seleccionamos que queremos 2 flotantes en el resultado
calculo = 5/17
print('El resultado es: {:.2f}'.format(calculo))
```

>**Función: F Strings**

```python
#Otra forma de realizar esto sin el .format es con las f Strings
nombre = "Adrian"
edad = 26
print(f"Mi nombre es {nombre} y tengo {edad} años" )
```
