Podemos abrir y leer el archivo que acabamos de crear

```python
mi_archivo = open('mi_archivo.txt')
```

```python
#Podemos ver que una vez se lee el archivo, nos escribirá una string vacía. Podemos reiniciar el cursor de lectura
mi_archivo.read()
```

```python
mi_archivo.seek(0)#Con seek aunque lo reiniciemos varias veces siempre recibiremos el texto
mi_archivo.read()
```

```python
mi_archivo.seek(0)
mi_archivo.readlines() #Lectura en formato multilinea
```

```python
#Tras el uso y para evitar errores necesitaremos cerrar el archivo
mi_archivo.close()
```

```python
#Escritura de archivos
#Tenemos que indicarle que será modo "w", escritura
f = open('mi_archivo.txt', "w")
f.write('Esto es una nueva linea')
f.close()
#No podríamos leer ya que está en modo escritura  -->  f.read()
f = open('mi_archivo.txt', "r")
f.read()


```

```python
#Podemos simplificar el proceso
with open('mi_archivo.txt', mode='r') as f:
    print(f.read())

with open('mi_archivo.txt', mode='a') as f:
    f.write('ANOTHER') 

with open('mi_archivo.txt', mode='r') as f:
    print(f.read())
```
