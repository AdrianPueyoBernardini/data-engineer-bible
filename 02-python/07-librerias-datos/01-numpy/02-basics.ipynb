## **Librerias datos: Numpy**

>**Basics**

```python
#Vamos a crear un array de 8 números con numpy:
import numpy as np
array = np.arange(8)
array
```

>**Vector VS Lista**

El vector, al ser un array de numpy nos permite hacer operaciones matemáticas con él de forma muy sencilla

```python
print(array +1) #Por ejemplo hacer un sumatorio de cada uno de los elementos del array anteriormente creado
print(array*3) #O multiplicar sus elementos
print(array/2) #O dividirlos
```

>**numpy: sum()**

```python
nuevo_array = array*5
print(nuevo_array)

#Sum nos permite sumar de forma super simple los elementos de un array
#Suma de array con un for
count = 0
for x in nuevo_array:
    count += x
print(count)

#funcion sum de numpy
count_2 = sum(nuevo_array)
print(count_2)
```

>**numpy: mean()**

```python
#mean nos da la media del array de forma simple
#Media de array con un for
count = 0
for x in nuevo_array:
    count += x

#count = count/nuevo_array.length()
count = count / len(nuevo_array)
print(count)

#funcion mean de numpy
count_2 = np.mean(nuevo_array)
print(count_2)
```
