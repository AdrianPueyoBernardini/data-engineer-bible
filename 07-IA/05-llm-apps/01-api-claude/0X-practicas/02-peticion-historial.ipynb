```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from anthropic import Anthropic
cliente = Anthropic()
modelo = "claude-sonnet-5-5"
```

```python
def save_user_msg(message, text):
    user_msg = {"role": "user", "content": text}
    message.append(user_msg)

def save_model_msg(message, text):
    model_msg = {"role": "assistant", "content": text}
    message.append(model_msg)

def chat(text):
    message = cliente.messages.create(
        model=modelo,
        max_tokens=500,
        messages=text
    )
    for block in message.content:
        if block.type == "text":
            return block.text
```

```python
mensajes = []

save_user_msg(mensajes, "Indicame el modelo de avión mas rápido en una frase")

respuesta = chat(mensajes)

save_model_msg(mensajes, respuesta)

save_user_msg(mensajes, "Ahora el siguiente")

respuesta_final = chat(mensajes)

save_model_msg(mensajes, respuesta_final)


for msg in mensajes:
    print(msg)
```
