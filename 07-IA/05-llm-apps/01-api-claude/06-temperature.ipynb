## Temperatura

La temperatura define como de determinista o creativo será nuestro modelo.   
*Especial atencion al nuevo valor **temperature** dentro de la función chat*

 ![api-claude](../../../images/api-claude-temp.png)

Depende de la tarea podemos ver que cada rango de temperatura es más adecuado o no.   

 ![api-claude](../../../images/api-claude-temp2.png)

```python

# 1. En esta función solicitamos la lista de mensajes totales y un texto para introducir un nuevo mensaje
# definimos el rol y contexto de forma rápida y añadimos con un append el mensaje nuevo a la lista total.
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

#2. Hacemos lo mismo con el mensaje de respuesta, como rol tendrá assistant
def add_assistant_message(messages, text):
    assistante_message = {"role": "assistant", "content": text}
    messages.append(assistante_message)

#3. Funcion automatizada en la que solicitamos la lista de mensajes y retornamos la respuesta habiéndo pasado el historial completo
def chat(messages, system, temperature=1.0):

    message = cliente.messages.create(
        system=system,
        model=model,
        max_tokens=1000,
        messages=messages,
        extra_body={"temperature": temperature}
    )
    for block in message.content:
        if block.type == "text":
            return block.text
```

```python
#Lista vacía de mensajes
messages = []

system="""
Eres un asistente de la aplicación filmia, da la bienvenida
"""

add_user_message(messages, "Dame una idea de pelicula de terror en una linea")
respuesta = chat(messages, system, 0.0)
add_assistant_message(messages, respuesta)

add_user_message(messages, "Dame otra más")
nueva_respuesta = chat(messages, system, 1.0)
add_assistant_message(messages, nueva_respuesta)
```

```python
line_counter = 1
for msg in messages:
    print(str(line_counter)+ ": " + msg['content'])
    line_counter+=1
```
