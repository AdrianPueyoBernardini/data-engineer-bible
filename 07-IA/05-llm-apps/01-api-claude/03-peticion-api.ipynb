```python
#Crear un cliente API
from anthropic import Anthropic

cliente = Anthropic()
model = "claude-sonnet-5-5"
```

```python
#Mensaje
message =cliente.messages.create(
    model = model,
    max_tokens = 500,
    messages=[
        {
            "role":"user",
            "content": "Defíneme ibón, responde en una frase corta"
        }
    ]
)
```

```python
print(message.content[0].text)
print(message.usage) 
```
