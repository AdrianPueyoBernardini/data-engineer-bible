```python
mensaje = input("Escribe tu pregunta: ")
```

```python
from anthropic import Anthropic
```

```python
cliente = Anthropic()
model = "claude-sonnet-5-5"
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
message = cliente.messages.create(
    model = model,
    max_tokens=500,
    messages=[
        {
            "role":"user",
            "content":mensaje
        }
    ]
)
```

```python
print(message.content[1].text)
print(message.usage) 
```
