## ¿Qué es la evaluación de prompts?

A diferencia del "Prompt engineering", el "Prompt Evaluation" lo que hace es generar flujos de evaluación de prompts para comprobar como de bien se han resuelto.   
Esta evaluación se puede definir en 5 pasos:
- 1. Hacer un borrador del prompt
```bash
    prompt = f"""
    Please answer the user's question:

    {question}
    """
```
- 2. Crear un dataset de evaluación, este contendrá pequeños ejemplos de preguntas o solicitudes que nuestro prompt resolverá en producción
- 3. Alimentar a Claude con las plantillas de preguntas + nuestro prompt
- 4. Cogeremos el prompt + la pregunta y la respuesta que nos ha dado Claude y lo pasaremos por el Grader para evaluar cada una de las respuestas.
- 5. Si el resultado es más bajo del esperado cambiamos nuestro Prompt y repetimos el proceso:   
    *Please answer the user's question: -> Please evaluate and answer precisely the user's question*
