## ¿Cómo hacer el prompt perfecto?

1. **Se claro y directo**
   - Usa lenguaje sencillo y facil de entender
   - Evita dar rodeos
   - No seas ambiguo, trata de ser muy específico
  
2. **A claude le ayuda usar XML para separar y diferenciar las partes del contenido.**
   **No es necesario utilizar etiquetas XML oficiales. Crea nombres descriptivos que tengan sentido para tu contenido:**   

      < athlete_information>   

         - Height: 6'2"
         - Weight: 180 lbs
         - Goal: Build muscle
         - Dietary restrictions: Vegetarian   
      
      </ athlete_information>   
   
3. **Utiliza ejemplos, varios ejemplos con diferentes escenarios es lo mejor si quieres enseñar diferentes respuestas válidas**

4. **Explica por que los ejemplos de respuesta utilizados se consideran ideales:**
