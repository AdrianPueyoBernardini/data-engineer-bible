# **Tools: Ejercicio recordatorio**

Vamos a realizarle la siguiente petición a claude:
> Pon un recordatorio para mi cita con el médico, es el jueves de la semana que viene.



**¿Por qué esto es un desafío?:**
- *Conciencia temporal limitada:* Claude puede saber la fecha actual, pero no la hora exacta.
- *Problemas con el cálculo de fechas:* Claude no maneja muy bien las sumas basadas en el tiempo, especialmente cuando se trata de calcular fechas con muchos días de anticipación.
- *Claude no sabe como configurar recordatorios ni tiene mecanismo para ello.*



**¿Qué herramientas necesitaremos?:**
- *Get the current date time*
- *Add duration to date time*
- *Set a reminder*


---

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system:
        params["System"] = system

    message = cliente.messages.create(**params)

    return message.content[0].text
```

```python
conversacion = []

add_user_message(conversacion, "Dame una idea para una aplicación móvil sencilla en máximo 2 líneas")

respuesta = chat(conversacion)

add_assistant_message(conversacion, respuesta)

add_user_message(conversacion, "Dame otra más")

respuesta= chat(conversacion)

add_assistant_message(conversacion, respuesta)
```
