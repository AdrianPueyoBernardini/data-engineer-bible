```python
import cv2

```

```python
#Upload a photo
from importlib.resources import files

from tkinter import Tk, filedialog

# Upload a photo
root = Tk()
root.withdraw()

photo_name = filedialog.askopenfilename(
    title="Select a photo",
    filetypes=[("Images", "*.jpg *.jpeg *.png")]
)

# Read the uploaded photo
img = cv2.imread(photo_name)


#Display the photo
cv2.imshow("Uploaded Photo", img)

#Display
print("Shape:", img.shape)

height = img.shape[0]
width = img.shape[1]
channels = img.shape[2]

#Aquí podemos ver los píxeles gracias a la librería cv2
print("Width:", width, "pixels")
print("Height:", height, "pixels")
print("Channels:", channels)
print("Total pixels", width * height)

# Read one pixel
pixel = img[50,50]

print("pixel at row 100, column 50 (B, G, R):", pixel)
```
